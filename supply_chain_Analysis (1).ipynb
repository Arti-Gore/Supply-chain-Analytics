# Key Insights

- Revenue was analyzed across different product types to understand product-wise performance.
- Product-wise sales volume was compared using the number of products sold.
- Supplier performance was analyzed using revenue, lead time, and defect rate.
- Transportation modes were compared based on usage and average shipping cost.
- Product quality was evaluated using average defect rates.
- Inventory performance was analyzed using average stock levels by product type.
- Manufacturing performance was studied using manufacturing costs and lead time.
- Shipping performance was analyzed using shipping time and shipping cost.
- Customer demographics and locations were analyzed to understand revenue and order patterns.
- Production volume, order quantity, and product sales were compared to understand supply chain demand and production patterns.

# Business Recommendations

- Focus on products and product categories that generate higher revenue and sales volume.
- Monitor supplier lead times to reduce delays in the supply chain.
- Review suppliers with relatively higher defect rates and improve quality control.
- Compare transportation modes and shipping costs to manage logistics expenses.
- Maintain appropriate stock levels to reduce the risk of stock shortages or excess inventory.
- Monitor manufacturing costs and lead times to improve production efficiency.
- Analyze shipping times regularly to identify opportunities to improve delivery performance.
- Use customer demographic and location insights for better demand planning.
- Align production volumes and order quantities with product demand.
